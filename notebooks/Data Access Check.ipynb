```python
#CELL 1 — Imports & load API key
import requests
import os
import json
from dotenv import load_dotenv

load_dotenv()
NVD_API_KEY = os.getenv("NVD_API_KEY")
BASE_URL = "https://services.nvd.nist.gov/rest/json/cves/2.0"

print(f"API Key loaded: {'✅' if NVD_API_KEY else '❌ NOT FOUND'}")
```

```python
# CELL 2 — Fetch a single CVE and see the raw response

headers = {"apiKey": NVD_API_KEY}
params = {
    "resultsPerPage": 5,
    "startIndex": 0
}

response = requests.get(BASE_URL, headers=headers, params=params)
print(f"Status code: {response.status_code}")

data = response.json()

# Pretty print the full raw response so we can see the shape
print(json.dumps(data, indent=2))
```

```python
# CELL 3 — Inspect top-level keys

print("Top-level keys:", list(data.keys()))
print("Total CVEs in NVD:", data["totalResults"])
print("Results returned:", data["resultsPerPage"])
```

```python
# CELL 4 — Drill into a single CVE object
cve = data["vulnerabilities"][0]["cve"]

print("CVE keys:", list(cve.keys()))
print("\nID:", cve["id"])
print("Published:", cve["published"])
print("Last Modified:", cve["lastModified"])
print("\nDescription:", cve["descriptions"][0]["value"])
```

```python
# CELL 5 — Inspect CVSS metrics
metrics = cve.get("metrics", {})
print("Available metric versions:", list(metrics.keys()))

if "cvssMetricV31" in metrics:
    cvss = metrics["cvssMetricV31"][0]["cvssData"]
    print("\nCVSS v3.1 Score:", cvss["baseScore"])
    print("Severity:", cvss["baseSeverity"])
    print("Attack Vector:", cvss["attackVector"])
    print("Attack Complexity:", cvss["attackComplexity"])
    print("Privileges Required:", cvss["privilegesRequired"])
    print("User Interaction:", cvss["userInteraction"])
    print("Confidentiality Impact:", cvss["confidentialityImpact"])
    print("Integrity Impact:", cvss["integrityImpact"])
    print("Availability Impact:", cvss["availabilityImpact"])
else:
    print("No CVSS v3.1 data on this CVE")
```

```python
# CELL 6 — Inspect CPE (affected products)
configurations = cve.get("configurations", [])
print(f"Configurations (CPE blocks): {len(configurations)}")

if configurations:
    print(json.dumps(configurations[0], indent=2))
```
